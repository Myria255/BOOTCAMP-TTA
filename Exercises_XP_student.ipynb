**To-Do:** Write your answers below.

### Answers:

1.  **What is the key difference between traditional machine learning and deep learning?**
    The key difference is in **feature extraction**. Traditional machine learning requires human experts to manually extract relevant features from raw data (feature engineering), while deep learning models automatically learn hierarchical features directly from raw data through multiple layers of abstraction. For example, in image classification, traditional ML might require hand-crafting features like edge detectors, whereas a deep learning model learns these features internally.

2.  **How do artificial neural networks mimic the human brain?**
    Artificial neural networks mimic the human brain by using **interconnected nodes (neurons)** organized in layers. Each neuron receives inputs, processes them, and passes the output to subsequent neurons, much like biological neurons transmit signals. This layered structure allows for the learning of complex patterns and representations.

3.  **Why does deep learning perform better on large datasets compared to traditional machine learning?**
    Deep learning models excel with large datasets because they have a **high capacity to learn complex patterns and representations** from vast amounts of data, which often contain more intricate relationships than smaller datasets. Traditional machine learning algorithms often hit a performance plateau as data increases, whereas deep learning models tend to improve with more data, leveraging its scale to learn more robust and generalizable features.

4.  **What are some challenges of deep learning, and how can they be addressed?**
    Some challenges of deep learning include **computational intensity**, requiring significant hardware (GPUs) and time for training, and **data dependency**, as they typically need very large datasets to perform well. They can also be prone to **overfitting** if the dataset is not diverse enough or the model is too complex. These can be addressed by using more efficient architectures, transfer learning with pre-trained models, data augmentation, regularization techniques (like dropout), and specialized hardware.

5.  **What is feature engineering, and why is it not needed in deep learning?**
    **Feature engineering** is the process of manually selecting and transforming raw data into features that can be understood by machine learning models, often requiring domain expertise. It is largely **not needed in deep learning** because deep neural networks, especially convolutional neural networks (CNNs) for image data or recurrent neural networks (RNNs) for sequential data, are designed to automatically learn relevant features from the raw input data itself through their layered architectures.

6.  **What role do hidden layers play in a deep learning model?**
    Hidden layers play a crucial role by **extracting and transforming complex patterns** from the input data into more abstract representations. Each hidden layer learns different levels of abstraction, with earlier layers identifying simpler features (e.g., edges in an image) and deeper layers combining these into more complex features (e.g., shapes or objects). This hierarchical learning allows deep learning models to understand intricate relationships in the data.

7.  **In an artificial neural network, what is the function of an activation function?**
    The function of an activation function is to **introduce non-linearity** into the neural network's output. Without activation functions, a neural network would simply be a linear regression model, regardless of the number of layers, limiting its ability to learn complex patterns and model non-linear relationships in the data. For example, the ReLU (Rectified Linear Unit) activation function outputs the input directly if it's positive, otherwise it outputs zero, allowing the network to learn more complex decision boundaries.

## 🌟 Exercise 5 (optional): Implementing Forward and Backward Propagation in Python

**As stated in the enoncé**  
You will code a simple neural network that performs forward propagation and backpropagation for a regression problem predicting exam scores from study hours.

**Guidance**  
Run the prefilled code. Then modify the learning rate or initial weights and observe changes in the loss and updates.
Provide a short written explanation.

**Learning point**  
Gradient descent updates parameters in the direction of steepest loss reduction.
The learning rate controls step size. Too small slows learning. Too large risks divergence.


```python
# PREFILLED: just execute
import numpy as np

# Initialize input data (features)
x = np.array([4, 80])  # 4 hours studied, previous test score: 80

# Initialize weights and bias
w = np.array([0.6, 0.3])  # Initial weights
b = 10  # Initial bias

# Forward Propagation
def forward_propagation(x, w, b):
    z = np.dot(x, w) + b  # Weighted sum
    return z  # Linear activation (No ReLU here, it's a regression task)

# Compute prediction
y_pred = forward_propagation(x, w, b)
y_true = 85  # Actual exam score

# Compute Loss (Mean Squared Error)
loss = 0.5 * (y_true - y_pred) ** 2

# Compute Gradients
grad_w = -(y_true - y_pred) * x  # Partial derivatives with respect to weights
grad_b = -(y_true - y_pred)  # Partial derivative with respect to bias

# Update Weights and Bias
learning_rate = 0.01
w_new = w - learning_rate * grad_w
b_new = b - learning_rate * grad_b

# Print Results
print("Initial Prediction:", y_pred)
print("Loss:", loss)
print("Updated Weights:", w_new)
print("Updated Bias:", b_new)
```
