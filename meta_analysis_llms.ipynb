# Mini-projet : Méta-analyse d’articles de recherche sur les LLM

**Thème choisi :** efficacité, alignement et raisonnement dans les grands modèles linguistiques ouverts.

Ce notebook présente une méta-analyse de cinq articles récents sur les Large Language Models. L’objectif n’est pas seulement de résumer les articles, mais aussi de comparer leurs contributions, leurs méthodologies, leurs limites et les tendances de recherche qui émergent.

## 1. Introduction

Les grands modèles linguistiques, ou **Large Language Models (LLM)**, sont des modèles d’intelligence artificielle entraînés sur de très grandes quantités de textes afin de comprendre, générer, résumer, traduire, coder et raisonner à partir du langage naturel. Depuis l’apparition de modèles comme GPT, LLaMA, Mistral ou DeepSeek, la recherche ne se limite plus seulement à augmenter la taille des modèles. Elle cherche aussi à rendre les modèles plus efficaces, plus alignés avec les préférences humaines et plus capables de résoudre des tâches complexes.

Cette méta-analyse se concentre sur un thème commun : **comment les LLM ouverts deviennent plus performants, plus efficaces et mieux alignés**. Les articles sélectionnés couvrent plusieurs dimensions complémentaires : l’architecture des modèles ouverts, l’optimisation mémoire pour le fine-tuning, l’alignement par préférences humaines et l’amélioration du raisonnement par apprentissage par renforcement.

Les articles analysés sont :

1. Grattafiori et al. (2024), *The Llama 3 Herd of Models*, arXiv.
2. Jiang et al. (2023), *Mistral 7B*, arXiv.
3. Dettmers et al. (2023), *QLoRA: Efficient Finetuning of Quantized LLMs*, NeurIPS 2023.
4. Rafailov et al. (2023), *Direct Preference Optimization: Your Language Model is Secretly a Reward Model*, NeurIPS 2023.
5. DeepSeek-AI et al. (2025), *DeepSeek-R1: Incentivizing Reasoning Capability in LLMs via Reinforcement Learning*, arXiv.

## 2. Résumés d’articles

### Article 1 — *The Llama 3 Herd of Models*

**Citation complète :** Grattafiori, A. et al. (2024). *The Llama 3 Herd of Models*. arXiv:2407.21783.  
**Lien :** https://arxiv.org/abs/2407.21783

Cet article présente **Llama 3**, une famille de modèles de fondation développée pour prendre en charge plusieurs capacités : multilinguisme, génération de code, raisonnement et usage d’outils. Le plus grand modèle présenté est un Transformer dense de **405 milliards de paramètres** avec une fenêtre de contexte pouvant atteindre **128 000 tokens**. L’article présente aussi des versions pré-entraînées et post-entraînées, ainsi que **Llama Guard 3** pour la sécurité des entrées et sorties.

Le problème de recherche traité est le suivant : comment construire une famille de LLM ouverts capable de rivaliser avec des modèles propriétaires avancés tout en restant exploitable par la communauté. La solution proposée repose sur une combinaison de pré-entraînement massif, de post-entraînement, d’évaluations empiriques étendues et de mécanismes de sécurité.

Les datasets exacts ne sont pas présentés comme un seul jeu de données téléchargeable. L’article décrit plutôt un pipeline d’entraînement à grande échelle. L’architecture est basée sur un **Transformer dense**. Les métriques d’évaluation couvrent plusieurs domaines : compréhension du langage, raisonnement, code, multilinguisme, sécurité et usage d’outils.

### Article 2 — *Mistral 7B*

**Citation complète :** Jiang, A. Q. et al. (2023). *Mistral 7B*. arXiv:2310.06825.  
**Lien :** https://arxiv.org/abs/2310.06825

Mistral 7B propose un modèle de seulement **7 milliards de paramètres**, mais optimisé pour obtenir de fortes performances avec un coût réduit. L’article montre que Mistral 7B dépasse Llama 2 13B sur les benchmarks évalués et dépasse Llama 1 34B sur certaines tâches de raisonnement, mathématiques et génération de code.

La contribution principale repose sur deux innovations : **Grouped-Query Attention (GQA)**, utilisée pour accélérer l’inférence, et **Sliding Window Attention (SWA)**, utilisée pour mieux traiter les longues séquences avec un coût réduit.

Le problème de recherche est l’efficacité : au lieu d’augmenter uniquement la taille du modèle, les auteurs cherchent à améliorer le rapport performance/coût. L’architecture est un **Transformer decoder-only** optimisé avec GQA et SWA. Les évaluations utilisent des benchmarks de compréhension, raisonnement, mathématiques, code et instruction-following.

### Article 3 — *QLoRA: Efficient Finetuning of Quantized LLMs*

**Citation complète :** Dettmers, T., Pagnoni, A., Holtzman, A., & Zettlemoyer, L. (2023). *QLoRA: Efficient Finetuning of Quantized LLMs*. NeurIPS 2023.  
**Lien :** https://arxiv.org/abs/2305.14314

QLoRA répond à un problème très concret : le fine-tuning des grands modèles coûte cher en mémoire GPU. Les auteurs proposent une méthode permettant de fine-tuner un modèle de **65 milliards de paramètres** sur un seul GPU de **48 Go**, tout en conservant des performances proches du fine-tuning classique en 16 bits.

La solution combine plusieurs innovations : quantification **4-bit NormalFloat (NF4)**, double quantification et paged optimizers. Le modèle de base reste gelé en 4 bits, tandis que l’apprentissage se fait à travers des adaptateurs **LoRA**. Les auteurs entraînent plus de 1 000 modèles sur plusieurs architectures et datasets d’instructions.

Leur meilleure famille de modèles, appelée **Guanaco**, atteint 99,3 % du niveau de performance de ChatGPT sur le benchmark Vicuna selon leur évaluation. Les architectures utilisées incluent notamment LLaMA et T5. Les métriques incluent des benchmarks de chatbot, des évaluations humaines et des évaluations par GPT-4.

### Article 4 — *Direct Preference Optimization*

**Citation complète :** Rafailov, R., Sharma, A., Mitchell, E., Manning, C. D., Ermon, S., & Finn, C. (2023). *Direct Preference Optimization: Your Language Model is Secretly a Reward Model*. NeurIPS 2023.  
**Lien :** https://arxiv.org/abs/2305.18290

DPO traite le problème de l’alignement des LLM avec les préférences humaines. Les méthodes classiques comme **RLHF** sont puissantes, mais elles sont souvent complexes : elles nécessitent d’entraîner un reward model, puis d’utiliser une procédure d’apprentissage par renforcement qui peut être instable et coûteuse.

La contribution principale de DPO est de simplifier ce processus. Les auteurs montrent qu’il est possible d’optimiser directement un modèle à partir de données de préférences humaines avec une perte de classification, sans entraîner séparément un reward model et sans utiliser une procédure RL complexe.

Les datasets utilisés concernent notamment la complétion de sentiment, le résumé et le dialogue simple. Les métriques évaluent la capacité du modèle à respecter les préférences humaines, à améliorer la qualité des réponses et à contrôler certains comportements. L’article montre que DPO peut égaler ou dépasser RLHF sur certaines tâches tout en étant plus simple à implémenter.

### Article 5 — *DeepSeek-R1*

**Citation complète :** DeepSeek-AI et al. (2025). *DeepSeek-R1: Incentivizing Reasoning Capability in LLMs via Reinforcement Learning*. arXiv:2501.12948.  
**Lien :** https://arxiv.org/abs/2501.12948

DeepSeek-R1 s’intéresse à un enjeu central : améliorer les capacités de raisonnement des LLM sans dépendre fortement de chaînes de raisonnement annotées par des humains. Les auteurs présentent **DeepSeek-R1-Zero**, entraîné par reinforcement learning à grande échelle sans supervised fine-tuning préalable, puis **DeepSeek-R1**, qui ajoute des données de démarrage à froid et un entraînement en plusieurs étapes.

La solution repose sur l’idée que le reinforcement learning peut faire émerger des comportements de raisonnement, comme l’auto-vérification, la réflexion et l’adaptation de stratégie. L’article indique que DeepSeek-R1 atteint des performances comparables à OpenAI-o1-1217 sur des tâches de raisonnement.

Les benchmarks incluent des tâches de mathématiques, programmation et raisonnement scientifique. Le modèle est aussi utilisé pour distiller des modèles plus petits basés sur Qwen et Llama, avec des tailles allant de 1,5B à 70B paramètres.

## 3. Analyse comparative

| Article | Objectif principal | Innovation clé | Architecture | Données / tâches | Évaluation | Forces | Limites |
|---|---|---|---|---|---|---|---|
| Llama 3 | Construire une famille de LLM ouverts très performants | Modèle dense jusqu’à 405B paramètres, contexte 128K, sécurité avec Llama Guard 3 | Transformer dense | Pré-entraînement et post-entraînement à grande échelle | Raisonnement, code, multilinguisme, sécurité, outil | Très complet, puissant, orienté usage réel | Coût d’entraînement élevé, transparence limitée sur certaines données |
| Mistral 7B | Obtenir une forte performance avec un petit modèle | GQA + Sliding Window Attention | Transformer decoder-only optimisé | Données non totalement détaillées ; tâches de raisonnement, code, instruction | Benchmarks NLP, code, math, humains et automatiques | Excellent rapport performance/coût | Moins détaillé sur les données d’entraînement |
| QLoRA | Réduire le coût mémoire du fine-tuning | Quantification 4-bit + LoRA | Modèles gelés + adaptateurs LoRA | Plusieurs datasets d’instructions | Vicuna, évaluations humaines, GPT-4 | Rend le fine-tuning accessible sur un seul GPU | Dépend de la qualité des données et des benchmarks |
| DPO | Simplifier l’alignement par préférences humaines | Optimisation directe sans reward model séparé | Applicable à plusieurs LLM | Données de préférences, résumé, dialogue, sentiment | Qualité des réponses, contrôle du comportement | Simple, stable, moins coûteux que RLHF | Dépend fortement de la qualité des préférences |
| DeepSeek-R1 | Améliorer le raisonnement complexe | RL pour faire émerger le raisonnement + distillation | LLM de raisonnement + modèles distillés | Math, code, STEM, données synthétiques | AIME, MATH, Codeforces, GPQA | Très prometteur pour le raisonnement | Risques de raisonnement long, lisibilité, sécurité et vérification encore difficiles |

Les articles montrent une évolution claire du domaine. **Llama 3** et **Mistral 7B** se concentrent sur l’architecture et la performance des modèles ouverts. **QLoRA** s’intéresse à l’accessibilité du fine-tuning. **DPO** améliore l’alignement avec les préférences humaines. **DeepSeek-R1** pousse plus loin la question du raisonnement, en montrant que certaines capacités peuvent émerger par apprentissage par renforcement.

Une différence importante concerne le coût. Llama 3 illustre une approche très grande échelle, difficilement reproductible pour de petites équipes. Mistral 7B et QLoRA montrent au contraire qu’il est possible d’obtenir des résultats intéressants avec moins de ressources. DPO et DeepSeek-R1 se situent dans la phase de post-training : ils cherchent à rendre les modèles plus utiles, mieux alignés et plus capables de raisonner.

## 4. Perspectives et réflexion

Plusieurs tendances émergent de cette méta-analyse.

Premièrement, la taille du modèle n’est plus le seul facteur de performance. Mistral 7B montre qu’une architecture bien optimisée peut rivaliser avec des modèles plus grands. Deuxièmement, l’efficacité devient un critère central : QLoRA rend le fine-tuning plus accessible, ce qui permet à des étudiants, chercheurs ou petites entreprises d’adapter des modèles sans infrastructure massive.

Troisièmement, l’alignement devient aussi important que la performance brute. Un modèle peut être puissant mais peu utile ou risqué s’il ne suit pas correctement les instructions humaines. DPO est donc important, car il simplifie l’alignement en supprimant plusieurs étapes complexes de RLHF.

Quatrièmement, le raisonnement devient un axe majeur de recherche. DeepSeek-R1 montre que l’apprentissage par renforcement peut encourager des comportements de réflexion et d’auto-vérification. Cette tendance pourrait transformer la manière dont les LLM résolvent des problèmes complexes en mathématiques, programmation, cybersécurité ou sciences.

Les limites communes concernent la transparence des données, la reproductibilité, la fiabilité des benchmarks et la sécurité. Plusieurs articles montrent de fortes performances, mais tous ne permettent pas de reproduire exactement l’entraînement. De plus, certains benchmarks peuvent être contaminés ou ne pas refléter les vrais usages. Enfin, les modèles ouverts favorisent l’innovation, mais peuvent aussi être détournés si les mécanismes de sécurité sont insuffisants.

Les futures recherches devraient donc se concentrer sur quatre priorités : améliorer la transparence des datasets, créer des benchmarks plus robustes, développer des modèles plus petits mais performants, et renforcer la sécurité des modèles ouverts.

## 5. Conclusion

Cette méta-analyse montre que la recherche sur les LLM ouverts avance dans plusieurs directions complémentaires. **Llama 3** représente la puissance des grands modèles ouverts modernes. **Mistral 7B** prouve que l’efficacité architecturale peut réduire la dépendance à la taille. **QLoRA** démocratise le fine-tuning en réduisant les besoins matériels. **DPO** simplifie l’alignement avec les préférences humaines. **DeepSeek-R1** ouvre une nouvelle voie vers des modèles capables de raisonner de manière plus autonome.

Globalement, le domaine évolue vers des modèles plus accessibles, plus efficaces, mieux alignés et plus spécialisés. Cependant, les défis restent importants : coût, sécurité, reproductibilité, fiabilité des évaluations et transparence. L’avenir des LLM ouverts dépendra donc de la capacité des chercheurs à équilibrer performance, responsabilité et accessibilité.
