```python
#EXERCISE1 

import pandas as pd

url = "https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv"
df = pd.read_csv(url)

prix_maisons = df['medv'].dropna()

moyenne = prix_maisons.mean()
mediane = prix_maisons.median()
ecart_type = prix_maisons.std()

print(f"Moyenne : {moyenne:}")
print(f"Médiane : {mediane:}")
print(f"Écart type : {ecart_type:}")
```

```python
#EXERCISE2

# Chargement des fichiers réels du dataset Retail fourni
sales = pd.read_csv("sales data-set.csv")
stores = pd.read_csv("stores data-set.csv")
features = pd.read_csv("Features data set.csv")

# 1. Nettoyage et préparation
sales['Date'] = pd.to_datetime(sales['Date'], format='%d/%m/%Y')

# 2. Analyse des tendances des ventes (Ventes hebdomadaires globales)
print("TENDANCES DES VENTES")
sales_trends = sales.groupby('Date')['Weekly_Sales'].sum().reset_index()
print(sales_trends.sort_values(by='Weekly_Sales', ascending=False).head())

# 3. Performance des magasins (Jointure avec les caractéristiques du magasin)
print("\n--- PERFORMANCE PAR TYPE DE MAGASIN ---")
sales_stores = pd.merge(sales, stores, on='Store', how='inner')
store_performance = sales_stores.groupby('Type')['Weekly_Sales'].agg(['sum', 'mean'])
print(store_performance)

# 4. Modèles d'achat (Impact des vacances / IsHoliday)
print("\n--- IMPACT DES JOURS FÉRIÉS SUR LES VENTES ---")
holiday_impact = sales.groupby('IsHoliday')['Weekly_Sales'].mean()
print(holiday_impact)
```

```python
#EXERCISE6 
!pip install Faker pandas

from faker import Faker
import random

# Initialisation de Faker (en français pour des noms et descriptions réalistes)
fake = Faker('fr_FR')

# Utilisation d'un set pour garantir l'unicité stricte des noms de produits
noms_produits = set()
while len(noms_produits) < 500:
    # Combinaison d'un adjectif et d'un mot pour créer des noms de produits uniques
    nom = f"{fake.word().capitalize()} {fake.color_name()}"
    noms_produits.add(nom)

liste_produits = list(noms_produits)
donnees_produits = []

# Génération des 500 produits
for i in range(500):
    produit = {
        "id_produit": f"PROD-{1000 + i}",  # Identifiant unique séquentiel
        "nom": liste_produits[i],          # Nom unique issu du set
        "description": fake.sentence(nb_words=12), # Description générée par Faker
        "prix": round(random.uniform(5.0, 250.0), 2) # Prix unique/aléatoire entre 5€ et 250€
    }
    donnees_produits.append(produit)

# Création du DataFrame Pandas
df_catalogue = pd.DataFrame(donnees_produits)

# Affichage des premières lignes pour vérification
print(df_catalogue.head())

```
