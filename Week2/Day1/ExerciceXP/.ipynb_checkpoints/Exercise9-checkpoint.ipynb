```python
import pandas as pd
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 32],
    'City': ['New York', 'Paris', 'Berlin', 'London']
}
df = pd.DataFrame(data)
```

```python
#Export to csv
df.to_csv('my_dataframe.csv', index=False)

# Export to JSON
df.to_json('my_dataframe.json')
```
