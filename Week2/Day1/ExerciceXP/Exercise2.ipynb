```python
#EXercise 1
## 1. What is Data Analysis?

#It is the process of cleaning, transforming, and modeling raw data to discover useful insights, draw conclusions, and support decision-making. It operates across four stages:

#Descriptive:** What happened?
#Diagnostic:** Why did it happen?
#Predictive:** What will happen?
#Prescriptive:** What should we do?


## 2. Why is it Critical Today?

#Volume & Velocity:** With the explosion of data (IoT, mobile apps, transactions), gut feeling is obsolete; businesses require fact-based decisions.
#Real-Time Action:** It enables immediate responses, such as blocking cyber threats or adjusting dynamic pricing instantly.
#Automation:** Paired with AI, it feeds closed-loop systems that automatically learn, adapt, and optimize workflows without human delays.


## 3. Three Key Areas of Application

### A. Cybersecurity & Finance (Fraud Detection)

#Application:** Real-time analysis of transaction logs and network traffic.
#Usage:** Instantly flagging anomalous behavior or baseline deviations to block fraudulent transactions and system intrusions before damage occurs.

### B. E-Commerce (Hyper-Personalization)

#Application:** Processing user clicks, hover times, and historical purchases.
#Usage:** Generating unique product recommendations and dynamically tailoring app interfaces to maximize user conversion.

### C. Smart Manufacturing (Predictive Maintenance)

#Application:** Monitoring IoT sensor telemetry (vibration, temperature) on industrial machinery.
#Usage:** Detecting micro-anomalies to predict equipment failure before it happens, minimizing costly factory downtime.

#EXercise 2
import pandas as pd

clean_dataset=pd.read_csv('clean_dataset.csv')
clean_dataset.head()

crx=pd.read_csv('crx.csv')
crx.head()

Mental=pd.read_csv('Mental.csv',low_memory=False)
Mental.head()
Time=pd.read_csv('Time.csv')
Time.head()
#Les données donnent des information sur des pays par rapport à leur habitude(alimentaire , consommation de stupefiant et etc)

#Debut exercise3
print(clean_dataset.dtypes)
print(Time.dtypes)
print(crx.dtypes)
print(Mental.dtypes)
#Les variables de type object sont classées comme qualitatives et les autres sont quantitatives

#EXercise 4

import os
import sys
import zipfile

!{sys.executable} -m pip install --quiet kaggle

os.environ['KAGGLE_CONFIG_DIR'] = "C:/Users/Debohi/Downloads/DI_Bootcamp/Week2/Day1/ExerciceXP"

from kaggle.api.kaggle_api_extended import KaggleApi
api = KaggleApi()
api.authenticate()

api.dataset_download_files('uciml/iris', path='.')

with zipfile.ZipFile('iris.zip', 'r') as zip_ref:
    zip_ref.extractall('.')

df_iris = pd.read_csv('Iris.csv')
df_iris.head()
df_iris.dtypes
# Le type object est une variable quantitative
#Les autres sont qualitatives


#EXercise 5


df = pd.read_csv("Time.csv")
df.columns = df.columns.str.strip()

print(df.columns.tolist())

sleep_col = next((col for col in df.columns if 'sleep' in col.lower() or 'duration' in col.lower()), None)

if sleep_col:
    print(df[sleep_col].describe())
    
    if 'Gender' in df.columns:
        print(df.groupby('Gender')[sleep_col].mean())
        
    if 'Year' in df.columns:
        print(df.groupby('Year')[sleep_col].mean())

#EXercise 6

#Financial reports of a company stored in an Excel file: Structured Data (or Semi-Structured Data depending on formatting).

#Photographs uploaded to a social media platform: Unstructured Data.

#A collection of news articles on a website: Unstructured Data (or Semi-Structured Data if considering metadata like dates and authors).

#Inventory data in a relational database: Structured Data.

#Recorded interviews from a market research study: Unstructured Data.

#Exercise 7

#1. Travel Blog Articles
#Method: Use Natural Language Processing (NLP) through Named Entity Recognition (NER) and Sentiment Analysis.

#Reasoning: Raw text contains hidden variables. NER automatically extracts specific entities (locations, dates, attractions), while sentiment analysis quantifies the reviewer's experience into numerical scores.

#Structured Output: A table with columns: [Article_ID], [Destination], [Travel_Date], [Activities], [Sentiment_Score].

#2. Customer Service Audio Calls
#Method: Apply Speech-to-Text Transcription followed by Text Mining and Keyword Classification.

#Reasoning: Audio files cannot be queried directly. Transcribing them converts the voice into text, allowing algorithms to filter calls by duration, detect customer emotion, and tag recurring issues.

#Structured Output: A database with columns: [Call_ID], [Duration_Sec], [Topic_Category] (e.g., Billing, Technical), [Resolution_Status] (Yes/No).

#3. Handwritten Brainstorming Notes
#Method: Use Intelligent Character Recognition (ICR/OCR) combined with Thematic Tagging.

#Reasoning: Handwriting requires digitization. ICR extracts the text from sketches or papers, and a simple classification matrix organizes the messy text fragments into predefined columns based on action items.

#Structured Output: A spreadsheet with columns: [Idea_ID], [Core_Concept], [Department_Owner], [Priority_Level] (High/Low).

#4. Cooking Video Tutorials
#Method: Use Multimodal Video Analysis (Computer Vision for object detection + Audio Transcription).

#Reasoning: A video contains sequential steps. Computer vision detects physical objects (ingredients, tools) on screen, while audio transcription extracts the spoken quantitative measurements (grams, minutes) and pairs them with specific video timestamps.

#Structured Output: A JSON object or relational table containing: [Recipe_Name], [Ingredients_List], [Tools_Needed], [Step_Number], [Timestamp].

#Exercise 8

import pandas as pd

train = pd.read_csv('train.csv')
train.head()

#Exercise 9

import pandas as pd
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 32],
    'City': ['New York', 'Paris', 'Berlin', 'London']
}
df = pd.DataFrame(data)
#Export to csv
df.to_csv('my_dataframe.csv', index=False)

# Export to JSON
df.to_json('my_dataframe.json')

#Exercice10

import io
import zipfile
import requests
import pandas as pd

url = "https://github.com/devtlv/Datasets-DA-Bootcamp-2-/raw/refs/heads/main/Week%204%20-%20Data%20Understanding/W4D3%20-%20Importing%20Data,%20Exporting%20D/posts.zip"
response = requests.get(url)

with zipfile.ZipFile(io.BytesIO(response.content)) as z:
    json_filename = z.namelist()[0]
    with z.open(json_filename) as f:
        df_json = pd.read_json(f)

df_json.head(5)
```
