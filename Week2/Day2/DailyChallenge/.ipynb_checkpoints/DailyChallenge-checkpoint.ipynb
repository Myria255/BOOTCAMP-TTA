```python
import pandas as pd

url = "https://raw.githubusercontent.com/devtlv/Datasets-DA-Bootcamp-2-/main/Week%204%20-%20Data%20Understanding/W4D4%20-%20Data%20Preprocessing%20%26%20T/ds_salaries.csv"
df = pd.read_csv(url)

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df["salary_normalized"] = scaler.fit_transform(df[["salary_in_usd"]])

from sklearn.decomposition import PCA

df_numeric = df.select_dtypes(include=["int64", "float64"])

pca = PCA(n_components=2)
df_pca = pca.fit_transform(df_numeric)

df["pca_1"] = df_pca[:, 0]
df["pca_2"] = df_pca[:, 1]

stats_experience = (
    df.groupby("experience_level")["salary_in_usd"]
    .agg(["mean", "median"])
    .reset_index()
)
print(stats_experience)
```
