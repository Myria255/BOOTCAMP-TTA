```python
#EXERCISE2

import statistics
from scipy import stats  
data = [12, 15, 13, 12, 18, 20, 22, 21]

mean = stats.gmean(data)

median = statistics.median(data)
ecart = statistics.stdev(data)
variance = statistics.variance(data)

print(f"Moyenne géo: {mean}\nMédiane: {median}\nÉcart-type: {ecart}\nVariance: {variance}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Définition des paramètres
mu = 50
sigma = 10
dis = norm(loc=mu, scale=sigma)

# 2. Création de l'axe des X (on va de mu - 3*sigma à mu + 3*sigma pour capter 99.7% de la courbe)
x = np.linspace(mu - 3 * sigma, mu + 3 * sigma, 500)

# 3. Calcul de la densité de probabilité (l'axe Y) à partir de ton objet 'dis'
y = dis.pdf(x)

# 4. Construction et affichage du graphique
plt.figure(figsize=(8, 5))
plt.plot(x, y, color='#E2583E', linewidth=2.5, label=f'Loi Normale ($\mu$={mu}, $\sigma$={sigma})')
plt.fill_between(x, y, color='#E2583E', alpha=0.15)  # Colore l'espace sous la courbe

# Habillage du graphique
plt.title('Visualisation de la Distribution Normale', fontsize=12, fontweight='bold')
plt.xlabel('Valeurs (X)')
plt.ylabel('Densité de probabilité')
plt.axvline(mu, color='gray', linestyle='--', label=f'Moyenne ($\mu$={mu})')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)

# Affichage de la fenêtre
plt.show()
```

```python
#EXERCICE4

data1 = np.random.normal(50, 10, 100)
data2= np.random.normal(60, 10, 100)

t_statistic, p_value = stats.ttest_ind(data1, data2)
print(f"t-statistic: {t_statistic:.4f}, p-value: {p_value:.4f}")
```

```python
#  EXERCISE5

house_sizes=[50, 70, 80, 100, 120]

house_prices=[150000, 200000, 210000, 250000, 280000]

#La pente et l'interpretation

slope, intercept, r_value, p_value, std_err = stats.linregress(house_sizes, house_prices)

#Prediction du prix d'une maison de 90m²

predicted_price = slope * 90 + intercept

print(f"Pente: {slope:.2f}, Intercept: {intercept:.2f}")
print(f"Prix prédit pour une maison de 90m²: {predicted_price:.2f}")

#interpretation de la pente
if slope > 0:
    print("La pente est positive, ce qui signifie que les prix des maisons augmentent avec la taille.")
elif slope < 0:
    print("La pente est négative, ce qui signifie que les prix des maisons diminuent avec la taille.")
else:
    print("La pente est nulle, ce qui signifie qu'il n'y a pas de relation linéaire entre la taille et le prix des maisons.")
```

```python
#EXERCISE6

fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

#Test ANOVA
f_statistic, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)
print(f"F-statistic: {f_statistic:.4f}, p-value: {p_value:.4f}")


#interpretation de la p-value
#Si la p-value est inférieure à 0.05, on rejette l'hypothèse nulle, ce qui suggère qu'il y a une différence significative entre les groupes de fertilisants. Sinon, on ne rejette pas l'hypothèse nulle, indiquant qu'il n'y a pas de différence significative entre les groupes.

#Interprétation
if p_value < 0.05:
    print("Il y a une différence significative entre les groupes de fertilisants.")
else:
    print("Il n'y a pas de différence significative entre les groupes de fertilisants.")
```
