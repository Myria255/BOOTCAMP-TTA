## Exercise 1: Rewrite and Optimize a Vague Prompt
**Scenario**: Manager asks for a LinkedIn post about FlowNest with the vague prompt `"Write something about productivity tips."`

### Tasks
- Identify 3 critical issues with the prompt.
- Rewrite it using: role prompting, clear instructions, and format/tone/length constraints.
- Add a final line to make it reusable for future social posts.

```python
# List three critical issues with the vague prompt
critical_issues = [
    "No clear target audience: the prompt does not say whether the post is for students, remote workers, busy professionals, or another group.",
    "No brand or product context: it does not mention FlowNest, its benefits, or the brand voice, so the answer may be generic.",
    "No platform or engagement guidance: LinkedIn content needs a hook, readable structure, professional tone, and a call to action.",
]
critical_issues
```

```python
# Rewrite the prompt using role prompting, clear instructions, and constraints
rewritten_prompt = """
Act as a senior social media copywriter for a productivity startup.

Write a LinkedIn post promoting our new focus app, FlowNest. The target audience is busy professionals, students, and remote workers who struggle with distractions and time management.

Requirements:
- Start with an engaging hook.
- Give 3 actionable productivity tips.
- Naturally explain how FlowNest helps users stay focused.
- Use a friendly, professional, and motivating tone.
- Keep the post between 120 and 160 words.
- Use short paragraphs or bullet points for readability.
- End with a clear call to action.
- Add no more than 3 relevant hashtags.

Use this format whenever we promote productivity tools on social media.
""".strip()

rewritten_prompt
```

## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
# Full multi-part prompt for educational quiz generation
quiz_prompt = """
Act as a 7th-grade science teacher creating educational content for students aged 11–13.

Using only the article provided below, create content about volcanic eruptions for Google Slides.

Output format:

Slide 1: Title Slide
- Create a simple and engaging title about volcanic eruptions.

Slide 2: Summary
- Summarize the article in exactly 2 simple bullet points.
- Use clear and age-appropriate vocabulary.

Slides 3–5: Quiz Questions
- Generate 3 multiple-choice questions.
- Each question must include 1 correct answer and 2 incorrect but believable distractors.
- Make the questions suitable for 7th-grade students.

Slide 6: Answer Key
- List the correct answer for each question.
- Add a short explanation for each correct answer.

Tone and language:
- Use a friendly and encouraging tone.
- Use simple vocabulary for 11–13-year-olds.
- Do not include information that is not found in the article.

Article:
[Paste the article here]
""".strip()

quiz_prompt
```

```python
# Explain three improvements over the vague prompt
improvements = [
    "It defines the audience clearly by specifying 7th-grade students aged 11–13, which helps the model choose age-appropriate vocabulary.",
    "It gives a precise Google Slides structure with a title slide, summary slide, quiz slides, and answer key, reducing messy or incomplete output.",
    "It restricts the model to the provided article, which helps prevent irrelevant information and hallucinated facts.",
]
improvements
```

## Exercise 4: Match Prompt to Purpose
Pick a style (Exploratory, Structured, Conversational, Functional) to build a customer-support chatbot prompt.

### Tasks
- Write the full prompt in the chosen style.
- Identify 2 features that prove it matches the style.
- Justify why this style fits better than the others.

```python
# Choose a style and write the customer-support chatbot prompt
chosen_style = 'Structured'

prompt_template = """
Act as a customer support chatbot for an e-commerce company.

When a customer sends a message, respond using this structure:

1. Greeting
Start with a polite and friendly greeting.

2. Issue Summary
Briefly restate the customer's issue in one sentence.

3. Solution or Next Steps
Provide clear steps to help the customer. Use a numbered list if there is more than one step.

4. Policy Reminder
If the issue involves delivery, returns, refunds, or damaged products, explain the relevant policy in simple language.

5. Closing
End with a helpful closing sentence and ask if they need anything else.

Rules:
- Keep the tone friendly and professional.
- Do not invent order details, delivery dates, refund status, or customer information.
- If important information is missing, ask one clear follow-up question.
- Keep the response short and easy to understand.

Customer message:
[Paste customer message here]
""".strip()

key_style_features = [
    "The prompt uses fixed response sections: greeting, issue summary, solution, policy reminder, and closing.",
    "It includes clear rules for tone, missing information, and avoiding invented customer details."
]

justification = (
    "The structured style is best for customer support because it creates consistent, clear, and reliable answers. "
    "Support responses must be easy to follow and should not forget key steps. Exploratory prompting would be too open-ended, "
    "conversational prompting may become too informal, and functional prompting may be too narrow for varied support cases."
)

chosen_style, prompt_template, key_style_features, justification
```

## Exercise 5: Prompt Refinement Challenge ? Control Style, Structure, and Length
**Scenario**: Write a prompt for PulseOne Mini smartwatch email blurb.

### Tasks
1) Enforce: friendly tone, bullet points, max 50 words, mention battery life, fitness tracking, Bluetooth.
2) After running, evaluate adherence and revise with stronger constraints if needed.

```python
# Initial prompt with strict constraints
pulseone_prompt_v1 = """
Act as a marketing copywriter for a wearable tech company.

Write a short product blurb for our new smartwatch, PulseOne Mini.

Strict requirements:
- Use a friendly and engaging tone.
- Use exactly 3 bullet points.
- Do not exceed 50 words total.
- Mention all 3 features: battery life, fitness tracking, and Bluetooth compatibility.
- Do not include an introduction or conclusion.
- Do not use more than one sentence per bullet point.
""".strip()

# Example result after running the prompt
pulseone_sample_output_v1 = """
- Long battery life keeps up with your day.
- Fitness tracking helps you follow steps, workouts, and goals.
- Bluetooth compatibility makes syncing and staying connected simple.
""".strip()

pulseone_prompt_v1, pulseone_sample_output_v1
```

```python
# Self-evaluate after running the prompt
sample_words = pulseone_sample_output_v1.replace('-', ' ').split()

required_features = ['battery life', 'fitness tracking', 'Bluetooth compatibility']
features_present = all(feature.lower() in pulseone_sample_output_v1.lower() for feature in required_features)

# The sample output is 25 words and includes all three required features.
eval_v1 = {
    'word_count': len(sample_words),
    'word_count_ok': len(sample_words) <= 50,
    'features_covered': features_present,
    'tone_ok': True,
    'notes': 'The output uses exactly 3 bullet points, stays under 50 words, mentions all required features, and has a friendly tone.',
}

eval_v1
```

```python
# Revised prompt with stronger constraint language
pulseone_prompt_v2 = """
Act as a marketing copywriter.

Write a product blurb for PulseOne Mini.

You must follow these rules exactly:
- Write exactly 3 bullet points.
- Do not exceed 50 words total.
- Each bullet must mention only one main feature.
- The 3 required features are: battery life, fitness tracking, and Bluetooth compatibility.
- Use a friendly and engaging tone.
- Do not add a title, introduction, conclusion, or extra explanation.

Before finalizing, check that the total word count is 50 words or fewer.
""".strip()

pulseone_prompt_v2
```

## Exercise 6: Hallucination Spotting and Mitigation
**Scenario**: Model added an unsupported claim in a climate/marine article summary.

### Tasks
1) Prompt the model to avoid hallucinations using only provided text.
2) Rewrite with verification language.
3) List 2 mitigation strategies.
4) Name 2 domains where hallucinations are risky and why.

```python
# Baseline anti-hallucination prompt
no_hallucination_prompt = """
Act as a careful research assistant.

Summarize the peer-reviewed article provided below about climate change and marine biodiversity.

Use only the information explicitly stated in the article. Do not add outside facts, predictions, statistics, or assumptions. If a detail is not mentioned in the article, do not include it.

Your summary should:
- Identify the main topic of the article.
- List the key findings.
- Include only claims supported by the text.
- Avoid exaggerated or unsupported statements.

Article:
[Paste the article here]
""".strip()

no_hallucination_prompt
```

```python
# Verification-focused rewrite
verification_prompt = """
Act as a careful academic summarizer.

Read the article below and produce a factual summary.

Verification rules:
- Only include claims that are explicitly stated in the article.
- Do not use outside knowledge.
- Do not invent statistics, dates, study results, or predictions.
- If the article does not provide a specific number or conclusion, write: "This information is not available in the article."
- If you are uncertain whether a claim is supported, leave it out.
- Keep the summary neutral and evidence-based.

Article:
[Paste the article here]
""".strip()

verification_prompt
```

```python
# List mitigation strategies and risky domains
mitigation_strategies = [
    "Source restriction: the prompt tells the model to use only the provided article and avoid outside knowledge.",
    "Uncertainty instruction: the prompt tells the model to state when information is not available instead of guessing."
]

risky_domains = {
    'domain_1': {
        'name': 'Healthcare',
        'why': 'Hallucinations can lead to incorrect medical advice, unsafe treatment decisions, or misunderstood health risks.'
    },
    'domain_2': {
        'name': 'Legal',
        'why': 'Hallucinations can create fake laws, fake case references, or wrong legal interpretations with serious consequences.'
    },
}

mitigation_strategies, risky_domains
```
