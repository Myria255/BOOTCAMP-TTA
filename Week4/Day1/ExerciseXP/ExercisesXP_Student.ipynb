```python

# This piece of code is already prefilled, run it to execute it and see the results.
# It demonstrates standard classification metrics for binary loan default prediction.

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix, classification_report

# Please replace these placeholders with your true labels and predicted probabilities.
y_true = [0,1,0,1,0,0,1,0,1,0]            # placeholder labels
y_pred_proba = [0.05,0.80,0.10,0.65,0.20,0.15,0.70,0.30,0.85,0.25]  # placeholder probabilities

# You should set a decision threshold that reflects the precision–recall trade-off for your business case.
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```

### Votre réponse

Pour l'**apprentissage supervisé**, nous évaluerions un modèle de classification en divisant d'abord les données en ensembles d'entraînement, de validation et de test (par exemple, 70% entraînement, 15% validation, 15% test). Une validation croisée K-fold (par exemple, K=5 ou K=10) serait utilisée sur l'ensemble d'entraînement pour obtenir des estimations robustes des performances du modèle et optimiser les hyperparamètres, tandis que l'ensemble de test serait réservé pour une évaluation finale non biaisée. Nous rapporterions des métriques telles que l'**exactitude (accuracy)**, la **précision (precision)**, le **rappel (recall)**, le **score F1**, la **courbe ROC (Receiver Operating Characteristic)** et l'**aire sous la courbe ROC (AUC-ROC)**, ainsi que la **courbe Précision-Rappel (PR-AUC)**, car elles offrent une vue complète des performances du modèle, notamment en présence de déséquilibre de classes. Le principal défi d'évaluation ici est le risque de surapprentissage (overfitting) ou de sous-apprentissage (underfitting) si la validation n'est pas effectuée correctement.

Pour l'**apprentissage non supervisé**, spécifiquement pour un modèle de clustering, nous évaluerions l'efficacité du regroupement sans étiquettes de vérité terrain. Nous mesurerions la qualité des clusters en utilisant des métriques intrinsèques comme le **score de silhouette**, qui quantifie la similarité d'un objet avec son propre cluster par rapport à d'autres clusters, et l'**indice de Davies-Bouldin**, qui mesure le rapport de la dispersion intra-cluster à la séparation inter-cluster. Pour déterminer le nombre optimal de clusters, la **méthode du coude (Elbow Method)** serait employée, en traçant la somme des carrés des erreurs intra-cluster (inertie) en fonction du nombre de clusters, cherchant le point où la diminution marginale de l'inertie devient moins significative. Des diagnostics visuels, tels que les **nuages de points avec les centroïdes** ou les **dendrogrammes** pour le clustering hiérarchique, seraient également utiles pour observer la structure des clusters. Le principal défi est l'absence de vérité terrain pour valider objectivement la qualité des regroupements.

Pour l'**apprentissage par renforcement**, nous mesurerions le succès du modèle en suivant la **récompense cumulative** (cumulative reward) de l'agent au fil du temps et sur plusieurs épisodes, cherchant une augmentation progressive indiquant un apprentissage efficace. La **convergence** serait évaluée en observant si la récompense cumulative ou la fonction de valeur de l'agent se stabilise après un certain nombre d'épisodes, suggérant que l'agent a appris une politique stable. L'équilibre entre l'**exploration** (découverte de nouvelles actions et états) et l'**exploitation** (utilisation des connaissances actuelles pour maximiser la récompense) serait géré par des stratégies comme la méthode epsilon-greedy, en ajustant le taux epsilon pour favoriser l'exploration au début et l'exploitation plus tard. Un défi majeur est la difficulté de définir une fonction de récompense appropriée qui guide l'agent vers le comportement souhaité sans introduire de biais indésirables.
