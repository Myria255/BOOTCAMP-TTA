# Daily Challenge: Precision Prompting for Control & Output Quality

This notebook completes the full prompt development cycle for the **Foundations of Prompt Engineering** daily challenge.

**Goal:** transform a dense internal security policy paragraph into clear employee-facing microlearning content while controlling tone, format, length, accuracy, and hallucination risk.


## Original Policy Text

> Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications.


## Step 1 — Prompt Controlling Tone, Format, and Length

### Prompt

You are an internal communications assistant for a global tech company. Rewrite the policy excerpt below into a friendly, clear knowledge snippet for employees.

Requirements:
- Use bullet points.
- Paraphrase the content; do not quote directly.
- Keep the tone friendly and professional.
- Stay under 75 words total.
- Keep all important security details from the source.

Policy excerpt:
"Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."


### Generated Response

- Always use the approved secure VPN when connecting to internal systems remotely.
- Do not use unsecured connections to access company data or sensitive messages.
- Avoid using personal devices that do not have endpoint protection.


```python
step1_output = '''- Always use the approved secure VPN when connecting to internal systems remotely.
- Do not use unsecured connections to access company data or sensitive messages.
- Avoid using personal devices that do not have endpoint protection.'''

word_count = len(step1_output.replace('-', ' ').split())
print('Step 1 word count:', word_count)
print('Under 75 words:', word_count < 75)
```

## Step 2 — Evaluate the Output

| Criteria | Guiding Question | Evaluation |
|---|---|---|
| Relevance | Does the output stick to the message of the source content? | Yes. It focuses only on secure remote access, VPN use, unsecured connections, personal devices, company data, and sensitive messages. |
| Clarity | Is the message easier to understand for non-technical staff? | Yes. The sentences are shorter and easier to scan than the original policy text. |
| Structure | Is the information formatted into clear bullet points? | Yes. The response uses three simple bullet points. |
| Tone | Is the language appropriate for internal employee communication? | Yes. The tone is friendly, direct, and professional. |
| Length | Does it stay under 75 words? | Yes. The response contains 34 words. |
| Factual Accuracy | Did the model make up or remove any critical security details? | No. It does not add new tools, policies, or technologies, and it keeps the key security requirements. |

### Overall Evaluation

The output is accurate, concise, and suitable for employee-facing microlearning content. It successfully simplifies the original paragraph without changing the core meaning.


## Step 3 — Detect and Mitigate Hallucinations

### Hallucination Check

The generated response does **not** appear to hallucinate. It does not add unrelated tools, new security rules, extra steps, or technologies that were not present in the source text.

However, to reduce hallucination risk even further, the prompt can be revised to strictly limit the model to the input.

### Revised Prompt

You are an internal communications assistant. Rewrite the policy excerpt below into a friendly, clear knowledge snippet for employees.

Strict requirements:
- Only paraphrase the information provided in the policy excerpt.
- Do not add new recommendations, tools, technologies, examples, or policy terms.
- Use bullet points.
- Avoid direct quotations.
- Stay under 75 words total.
- Keep every critical security detail: approved secure VPN, internal systems, unsecured connections, personal devices without endpoint protection, proprietary data, and sensitive communications.

Policy excerpt:
"Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."


## Step 4 — Paraphrasing Deep Dive for a Junior Intern Audience

### Prompt

Rewrite the policy paragraph for a junior intern audience using plain language and short phrases.

Requirements:
- Use no more than 4 bullet points.
- Avoid corporate or legal jargon.
- Keep the tone supportive and informative.
- Only use information from the input text.
- Do not add extra rules, tools, or advice.

Policy paragraph:
"Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."


### Generated Response for Junior Interns

- Working remotely? Use the approved secure VPN.
- Do not use unsafe connections for company systems.
- Do not access company data or sensitive messages from a personal device that is not protected.
- Keep remote access safe and approved.


```python
step4_output = '''- Working remotely? Use the approved secure VPN.
- Do not use unsafe connections for company systems.
- Do not access company data or sensitive messages from a personal device that is not protected.
- Keep remote access safe and approved.'''

bullet_count = sum(1 for line in step4_output.splitlines() if line.strip().startswith('-'))
word_count = len(step4_output.replace('-', ' ').split())
print('Step 4 bullet count:', bullet_count)
print('Step 4 word count:', word_count)
print('No more than 4 bullet points:', bullet_count <= 4)
```

## Step 5 — Quote Extraction Variant

### Prompt

Extract one direct quote from the original policy text that best captures the core security policy.

Requirements:
- Return only one direct quote.
- Do not paraphrase.
- Do not add explanation before or after the quote.
- Choose the quote that best communicates the main rule employees must follow.

Original policy text:
"Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."


### Extracted Quote

> "Employees must ensure that all remote access to internal systems is established via the approved secure VPN."


### When would quoting be more appropriate than paraphrasing?

Quoting is more appropriate in formal internal communications where exact wording matters, such as:

- Official policy updates
- Compliance reminders
- Security training materials
- Audit-related communications
- Messages from Legal, Compliance, IT, or Security teams

In these cases, using the exact policy wording helps avoid confusion and shows employees the official rule.

### When might quoting pose a risk?

Quoting can pose a risk when the original text is too dense, formal, or technical for the audience. Employees may misunderstand the rule if there is no plain-language explanation. Quoting may also be risky if the quote is taken out of context, becomes outdated, or leaves out other important policy details.
