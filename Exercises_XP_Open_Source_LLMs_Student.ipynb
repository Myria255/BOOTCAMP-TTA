```

# Exercise 1 — Comparison table
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Architecture, code, weights, training/fine-tuning pipeline and documentation | Best option for deep modification, auditing, domain adaptation and retraining. |
| Weights Released | Pretrained weights and usually inference examples/model card | Good for inference and fine-tuning, but harder to fully reproduce or audit the original training. |
| Architecture Only | Model design, paper or architecture description | Useful for research, but not practical for fast deployment because pretrained weights are missing. |"""
print(comparison_table)
```

## Exercise 4: Local Readiness Audit
**Use:** RAM/disk/OS; llama.cpp requirements (AVX/SSE, cmake/make/gcc/clang); quantized formats (GGUF).

**Step-by-step**
1) Gather system specs (RAM, free disk, OS type/version).  
2) Fill the audit table and mark ?/?.  
3) Check llama.cpp readiness (instruction sets, compilers, build tools).  
4) Identify upgrade needs for each ?.

**Deliverables:** filled table + upgrade summary.

```

# Exercise 4 — Gather system specs automatically
# Run this cell on the machine you want to audit. If you run it in Colab, it audits the Colab runtime.
import os, platform, shutil, subprocess, sys

try:
    import psutil
except ImportError:
    import subprocess, sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "psutil", "-q"])
    import psutil

ram_gb = round(psutil.virtual_memory().total / (1024**3), 2)
free_disk_gb = round(shutil.disk_usage("/").free / (1024**3), 2)
os_name = f"{platform.system()} {platform.release()}"

system_specs = {
    "ram_gb": ram_gb,
    "free_disk_gb": free_disk_gb,
    "os": os_name,
}
system_specs
```

```

# Exercise 4 — Readiness table
ram_ok = system_specs["ram_gb"] >= 16
storage_ok = system_specs["free_disk_gb"] >= 40
os_ok = ("Linux" in system_specs["os"]) or ("WSL" in system_specs["os"]) or ("Microsoft" in platform.platform())

readiness_table = f"""| Requirement | Your System Specs | Meets Requirement? |
| --- | --- | --- |
| RAM (>= 16 GB) | {system_specs['ram_gb']} GB | {'✅' if ram_ok else '❌'} |
| Free Disk Space (>= 40 GB) | {system_specs['free_disk_gb']} GB | {'✅' if storage_ok else '❌'} |
| OS (Linux/WSL2) | {system_specs['os']} | {'✅' if os_ok else '❌'} |"""
print(readiness_table)
```

```

# Exercise 4 — llama.cpp readiness notes
# llama.cpp commonly uses CMake and a C/C++ compiler. CPU instruction support such as AVX/AVX2/SSE can improve x86 performance.

def get_cpu_flags():
    if os.path.exists("/proc/cpuinfo"):
        with open("/proc/cpuinfo", "r", encoding="utf-8", errors="ignore") as f:
            text = f.read().lower()
        flags = []
        for flag in ["avx512", "avx2", "avx", "sse4_2", "sse4_1", "sse2", "neon"]:
            if flag in text:
                flags.append(flag)
        return flags or ["No common flags detected in /proc/cpuinfo"]
    return ["Check CPU flags manually on Windows: Task Manager / CPU-Z / systeminfo"]

def command_exists(cmd):
    return shutil.which(cmd) is not None

llama_cpp_readiness = {
    "cpu_instruction_support": get_cpu_flags(),
    "tooling": {
        "cmake_installed": command_exists("cmake"),
        "make_installed": command_exists("make"),
        "gcc_installed": command_exists("gcc"),
        "clang_installed": command_exists("clang"),
    },
    "other_requirements": "Use GGUF quantized models for local inference; for Windows, WSL2 or Visual Studio C++ build tools are recommended."
}
llama_cpp_readiness
```

```

# Exercise 4 — Upgrade / mitigation actions
upgrade_actions = []

if not ram_ok:
    upgrade_actions.append("RAM: upgrade to at least 16 GB for comfortable 7B quantized inference; for 8 GB, prefer 1B–3B models or very small GGUF quantizations.")
else:
    upgrade_actions.append("RAM: sufficient for many 7B quantized models, depending on quantization level and context length.")

if not storage_ok:
    upgrade_actions.append("Disk: free at least 40 GB for model files, environments and cache; remove old files or add external/extra storage.")
else:
    upgrade_actions.append("Disk: sufficient for several quantized models and tooling.")

if not os_ok:
    upgrade_actions.append("OS: install Linux or enable WSL2 on Windows before building/running llama.cpp.")
else:
    upgrade_actions.append("OS: compatible with a Linux/WSL2-style llama.cpp workflow.")

if not (llama_cpp_readiness["tooling"]["cmake_installed"] and (llama_cpp_readiness["tooling"]["gcc_installed"] or llama_cpp_readiness["tooling"]["clang_installed"])):
    upgrade_actions.append("Tooling: install CMake and a C/C++ compiler such as gcc, clang or Visual Studio C++ Build Tools.")
else:
    upgrade_actions.append("Tooling: compiler and CMake setup looks ready for llama.cpp builds.")

upgrade_actions
```
